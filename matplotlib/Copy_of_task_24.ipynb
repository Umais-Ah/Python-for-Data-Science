### **Problem-1:** Make a subplots which have 2 plots.

- For the first chart, draw a scatter plot "*Monitored Cap.(MW)*" vs "*Total Cap. Under Maintenance (MW)*" of top 5 most frequent power stations. Then draw the lines which indicate the average values of these two columns. Change the colors according to the names of the Power Stations.
- For the second chart, draw a scatter plot "*Monitored Cap. (MW)*" vs "*Actual(MU)*" of the top 5 most frequent power stations. Also draw the lines which indicates the average values of these two columns. Change the colors according to the names of the Power Stations.


````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df=pd.read_csv('/content/PowerGeneration.csv')
df.head()


````

````python
fig,ax=plt.subplots(nrows=1,ncols=2,figsize=(15,7))

temp=df[df['Power Station'].isin(df['Power Station'].value_counts().head().index[:])]
temp['Power Station']=temp['Power Station'].replace({ 'APL':0, 'NHPC':1, 'NPCIL':2, 'NTPC Ltd.':3, 'GMR ENERG':4})
p=ax[0].scatter(temp['Monitored Cap.(MW)'],temp['Actual(MU)'],c=temp['Power Station'],cmap='winter')

ax[0].axvline(temp['Monitored Cap.(MW)'].mean())
ax[0].axhline(temp['Actual(MU)'].mean())
fig.colorbar(p)
ax[0].set_xlabel('Monitored Cap')
ax[0].set_ylabel('Actual Mu')
ax[0].set_title('Monitored Cap vs Actual Mu')


p=ax[1].scatter(temp['Monitored Cap.(MW)'],temp['Total Cap. Under Maintenace (MW)'],c=temp['Power Station'],cmap='winter')
ax[1].axvline(temp['Monitored Cap.(MW)'].mean())
ax[1].axhline(temp['Total Cap. Under Maintenace (MW)'].mean())
fig.colorbar(p)
ax[1].set_xlabel('Monitored Cap')
ax[1].set_ylabel('Total Cap. Under Maintenace')
ax[1].set_title('Monitored Cap vs Total Cap. Under Maintenace')


````

### **Problem-3:** Make a 3D *Surface* plot of this below mathematical equation.

$$z = |x| - |y|$$

````python
x=np.linspace(-10,10,100)
y=np.linspace(-10,10,100)
xx,yy=np.meshgrid(x,y)
z=abs(xx)-abs(yy)
fig=plt.figure()
ax=plt.subplot(projection='3d')
p=ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
````

### `Problem-6` Use Pandas plot functions

* Line plot of closing value of top 5 Stocks in Year 2020.
* Take top 5 stocks based on total turnover in Year 2020

````python
df=pd.read_csv('/content/nifty-50.csv')
df['Date']=pd.to_datetime(df['Date'])
df.head()
````

````python
# 1. Reset/Load data and convert Date to datetime correctly
# 2. Filter for Year 2020
temp = df[df['Date'].dt.year == 2020]

# 3. Get top 5 stocks based on total turnover in 2020
top_5 = temp.groupby('Symbol')['Turnover'].sum().index[:5]

# 4. Filter data for just those 5 stocks
top_5_df = temp[temp['Symbol'].isin(top_5)]

# 5. Create the pivot table (handles duplicate dates automatically)
plot_data = top_5_df.pivot_table(index='Date', columns='Symbol', values='Close', aggfunc='mean')
plot_data
# 6. Line plot using Pandas
plot_data.plot(figsize=(12, 6), title='Daily Closing Value of Top 5 Stocks by Turnover (Year 2020)')
````

### `Problem-7` Scatter plot Close price vs Volume for TOP-5 Stocks in year 2021
* Color on Symbol Column

````python
temp = df[df['Date'].dt.year == 2021]
top_5=temp.groupby('Symbol')['Turnover'].sum().sort_values(ascending=False).index[:5]
top_5
top_5_df = temp[temp['Symbol'].isin(top_5)]
top_5_df['Symbol']=top_5_df['Symbol'].astype('category')
top_5_df.plot(kind='scatter',x='Close',y='Volume',c='Symbol',cmap='viridis')
````
