```python
# from a pandas dataframe
batsman=pd.read_csv('/content/sharma-kohli.csv')
batsman

plt.plot(batsman['index'],batsman['V Kohli'])
```

```python
# plotting multiple plots
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
```

```python
# labels title

plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
plt.title('Rohit Sharma  vs Virat Kohli Career Comparision')
plt.xlabel('Season')
plt.ylabel('Runs Scrored')
```

```python
# colors(hex) and line(width and style) and marker(size)

plt.plot(batsman['index'],batsman['V Kohli'],color='green')
plt.plot(batsman['index'],batsman['RG Sharma'],color='#F54927')
plt.title('Rohit Sharma  vs Virat Kohli Career Comparision')
plt.xlabel('Season')
plt.ylabel('Runs Scrored')
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='dashed')
# plt.plot(batsman['index'],batsman['RG Sharma'],color='#F54927',linestyle='dotted')
plt.plot(batsman['index'],batsman['RG Sharma'],color='#F54927',linestyle='dashdot')

plt.title('Rohit Sharma  vs Virat Kohli Career Comparision')
plt.xlabel('Season')
plt.ylabel('Runs Scrored')
```

```python

plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='solid',linewidth=3,marker='D',markersize=10)
# plt.plot(batsman['index'],batsman['RG Sharma'],color='#F54927',linestyle='dotted')
# plt.plot(batsman['index'],batsman['RG Sharma'],color='#F54927',linestyle='dashdot',linewidth=2,marker='D')
# plt.plot(batsman['index'],batsman['RG Sharma'],color='#F54927',linestyle='dashdot',linewidth=2,marker='+')
# plt.plot(batsman['index'],batsman['RG Sharma'],color='#F54927',linestyle='dashdot',linewidth=2,marker='>')
# plt.plot(batsman['index'],batsman['RG Sharma'],color='#F54927',linestyle='dashdot',linewidth=2,marker='.')
plt.plot(batsman['index'],batsman['RG Sharma'],color='#F54927',linestyle='dashdot',linewidth=2,marker='o')



plt.title('Rohit Sharma  vs Virat Kohli Career Comparision')
plt.xlabel('Season')
plt.ylabel('Runs Scrored')
```

```python
# legend -> location


plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='solid',linewidth=3,marker='D',markersize=10,label='Virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color='#F54927',linestyle='dashdot',linewidth=2,marker='o',label='Rohit')



plt.title('Rohit Sharma  vs Virat Kohli Career Comparision')
plt.xlabel('Season')
plt.ylabel('Runs Scrored')

plt.legend()#loc=best by default->automatically decide
# plt.legend(loc='right')

  #other options
  #  best',
  # 'upper
  #  right',
  # 'upper left'
  # 'lower left'
  # 'lower right'
  # 'right'
  # 'center left'
  # 'center right'
  # 'lower center'
  # 'upper center'
  # 'center'
```

```python
# limiting axes

price=[48000,54000,57000,49000,47000,45000,450000000]#because of outlier the other curve look flat
year=[2015,2016,2017,2018,2019,2020,2021]

plt.plot(year,price)
#we can limit x or y
plt.ylim(0,75000)
plt.xlim(2017,2019)
```

```python
# grid
plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='solid',linewidth=3,marker='D',markersize=10,label='Virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color='#F54927',linestyle='dashdot',linewidth=2,marker='o',label='Rohit')
plt.title('Rohit Sharma  vs Virat Kohli Career Comparision')

plt.grid()#enable grid ->improve readibility

plt.show()#have to call this in other text editor to make graph visible
```

```python
# plt.scatter on pandas data
df=pd.read_csv('batter.csv')
df=df.head(50)
```

```python
plt.scatter(df['avg'],df['strike_rate'],color='red',marker='+')
plt.title('Avg and Sr analysis of top 50 batsman')
plt.xlabel('Average')
plt.ylabel('Strike Rate')
```

```python
# size

tips=sns.load_dataset('tips')


plt.scatter(tips['total_bill'],tips['tip'],s=tips['size']*20)#dot size represent size of peoples
# slower

```

### Pie Chart

image.png

- Univariate/Bivariate Analysis
- Categorical vs numerical
- Use case - To find contibution on a standard scale

```python
# simple data
data=[23,45,100,20,49]
subjects=['eng','science','maths','sst','urdu']
plt.pie(data,labels=subjects)
plt.show()
```

```python
# dataset

df=pd.read_csv('/content/gayle-175.csv')
df
```

```python
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%')
plt.show()
```

```python
# percentage and colors

plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%',colors=['blue','purple','cyan','brown','pink','yellow'])
plt.show()

```

```python
# explode shadow

plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%',explode=[0.1,0,0,0,0,0],shadow=True)
plt.show()
```
