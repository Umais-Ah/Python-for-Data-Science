### **`Problem-1:`** Draw a line plot of which, the x-axis is the "Year" and the y-axis is sum of "PM2.5" of two countries Iran and China.

```python
# code here
df=pd.read_csv('https://tinyurl.com/2fe6vz4u')
iran_series=df.query('Country=="Iran"').groupby('Year')['PM2.5'].sum()
china_series=df.query('Country=="China"').groupby('Year')['PM2.5'].sum()
china_series
iran_series
```

```python
plt.plot(iran_series.index,iran_series.values,label='Iran')
plt.plot(china_series.index,china_series.values,label='China')
plt.legend()
plt.xlabel('Year')
plt.ylabel('PM2.5')
plt.xticks(df['Year'].unique())
plt.grid()
plt.title('PM2.5 of Iran and China over years')
plt.show()
```

###`Problem-6`
Show a line plot of Total Profit for each month with below styling.
* Dotted Line
* Line Color Blue

```python
# code here
df=pd.read_csv('https://docs.google.com/spreadsheets/d/e/2PACX-1vTJh6X4_mqixWsfK9mgkllGQkKYW9Wj9kOIMGY2uYsWeS8n5np87DO-SDGQWJ1HXEnxiOVFVzYFYEcR/pub?gid=558678488&single=true&output=csv')
plt.plot(df['month_number'],df['total_profit'],color='blue',linestyle='dotted',label='Total Profit',marker='o')
plt.legend(loc='upper left')
plt.grid()
plt.xlabel('Months')
plt.ylabel('Total Profit')
plt.title('Total Profit of Company Each Month')
```

###`Problem-7`
Show sales of each product in march month as pie chart.
* Show Percentage value
* Give Title "Sales in March"
* Explode ToothPaste with shadow

```python
# code here
products=df.query('month_number==3').stack().iloc[1:-2].index.get_level_values(1)
values=df.query('month_number==3').stack().iloc[1:-2].values
plt.pie(values,labels=products,autopct='%0.1f%%',explode=[0,0,0.1,0,0,0],shadow=True)
plt.title("Sales in March")
# plt.show()
```

###`Problem-9` Show Quarter wise Sales data for all products as multi Bar chart.


```python
# code here
df['date']=pd.to_datetime([f'2026-{month}-01' for month in df['month_number']])
df
```

```python
final_df=df.groupby(df['date'].dt.quarter)[['facecream', 'facewash', 'toothpaste', 'bathingsoap', 'shampoo', 'moisturizer']].sum()
final_df
```

```python
i=-1
for col in final_df.columns:
    plt.bar([a + i for a in final_df.index],final_df[col],width=0.15,label=col)
    i=i-0.15

plt.xlabel('Product')
plt.legend()
plt.ylabel('Sales')
plt.xticks(final_df.index-1.4,final_df.index)
plt.show()
```
