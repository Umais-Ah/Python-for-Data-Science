###`Q-8:` Player with most appearance in final match.

`Team1Players` and `Team2Players` have all players name. It is not a list of players name instead it is str. So handle it as string.

Hint: split and strip will help; Make a series of all players in final and do value counts


````python
# Assuming 'ipl' is your dataframe
# 1. Filter for only the 'Final' matches
final_matches = ipl[ipl['MatchNumber'] == 'Final']

# 2. Combine the two player columns into one long string
# We use a space to join them so we can easily split them later
all_players_str = (final_matches['Team1Players'] + ', ' + final_matches['Team2Players'])

# 3. Process the strings: remove brackets/quotes and split into a list
# This creates a flat list of all players who appeared in all final matches
all_players_list = []
for entry in all_players_str:
    # Remove unwanted characters and split by comma
    players = entry.replace('[', '').replace(']', '').replace("'", "").split(',')

    for player in players:
        all_players_list.append(player.strip())

# 4. Use value_counts to find the player with the most appearances
player_counts = pd.Series(all_players_list).value_counts()

# The result is the top player
player_counts.size
print(player_counts)
````

###`Q-9:` IPL Point Table

Make a function `point_table` which take `season` as parameter and show points table in non-ascendng order of points and in ascending order of team name.

For winning - 2 Ponits;
For loosing - 0 Point
For not result both team gets 1 points.

Make dataframe which will have
`TeamName` `MatchesPlayed` `MatchesWon` `NoResult` `Points`
make `TeamName` as index.

```
season parametr should be one of these->
['2022', '2021', '2020/21', '2019', '2018', '2017', '2016', '2015', '2014', '2013', '2012', '2011', '2009/10', '2009', '2007/08']
```


Output of two Top 2 in season 2022
```
TeamName    MatchesPlayed	MatchesWon	NoResult	Points

Gujarat Titans	    16	   12	       0	     24
Rajasthan Royals	  17	   10	       0	     20

```

````python
ipl.head(1)
````

````python
def Point_Tables(season):

    teams = ipl['Team1'][ipl['Season'] == season].unique()
    data = []

    for t in teams:

        match_played = ipl[((ipl['Team1'] == t) | (ipl['Team2'] == t)) & (ipl['Season'] == season)].shape[0]

        match_won = ipl[(ipl['WinningTeam'] == t) & (ipl['Season'] == season)].shape[0]

        no_result = ipl[((ipl['Team1'] == t) | (ipl['Team2'] == t)) &
                        (ipl['Season'] == season) &
                        (ipl['WinningTeam'].isna())].shape[0]

        points = match_won * 2 + no_result * 1

        data.append([t, match_played, match_won, no_result, points])

    df = pd.DataFrame(data, columns=['Team','MatchesPlayed','MatchesWon','NoResult','Points'])

    df = df.sort_values(by=['Points','Team'], ascending=[False, True])

    df.set_index('Team', inplace=True)

    return df


point_table = Point_Tables('2022')
point_table
````
