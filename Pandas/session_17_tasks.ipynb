Consider the following Python dictionary data and Python list labels:

```
data = {'birds': ['Cranes', 'Cranes', 'plovers', 'spoonbills', 'spoonbills', 'Cranes', 'plovers', 'Cranes', 'spoonbills', 'spoonbills', 'Cranes'],
```

###`Q-1:`
i. Create a DataFrame birds from the above dictionary data which has the index labels.

ii. Display basic information about the dataFrame.

iii. Show Alternate rows of the dataframe.

````python

import numpy as np
import pandas as pd

data = {
    'birds': ['Cranes', 'Cranes', 'plovers', 'spoonbills', 'spoonbills', 'Cranes', 'plovers', 'Cranes', 'spoonbills', 'spoonbills', 'Cranes'],
    'age': [3.5, 4, 1.5, np.nan, 6, 3, 5.5, np.nan, 8, 4, 3.5],
    'visits': [2, 4, 3, 4, 3, 4, 2, 2, 3, 2, 2],
    'priority': ['yes', 'yes', 'no', np.nan, 'no', 'no', 'no', 'yes', 'no', 'no', 'yes']
}
labels = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j', 'k']

# i. Create DataFrame with labels index
birds = pd.DataFrame(data, index=labels)

# ii. Display basic info
print("--- Basic Info ---")
birds.info()

# iii. Show alternate rows
print("\n--- Alternate Rows ---")
birds.iloc[::2]
````

###`Q-2:`
i. Show only rows [1st, 3rd, 7th] from columns ['bird', 'age']

ii. Select rows where the number of visits is less than 4.

````python
# i. Show only rows [1st, 3rd, 7th] from columns ['birds', 'age']
# Converting row numbers to 0-based positions (0, 2, 6)

print(birds.iloc[[0, 2, 6],[0,1]])

# ii. Select rows where the number of visits is less than 4
print("\n--- Visits < 4 ---")
birds[birds['visits'] < 4]
````

###`Q-3:`
i. Select all rows with nan values in age and visits column.

ii. Fill nan with respective series mode value.

````python
birds[birds['age'].isna() | birds['priority'].isna()]
birds.loc[:, 'age'] = birds.loc[:, 'age'].fillna(birds.loc[:, 'age'].mode()[0])
birds.loc[:, 'priority'] = birds.loc[:, 'priority'].fillna(birds.loc[:, 'priority'].mode())
birds.head()
````

###`Q-4`
i. Find the total number of visits of the bird Cranes

ii. Find the number of each type of birds in dataframe.

iii. Print no of duplicate rows

iv. Drop Duplicates rows and make this changes permanent. Show dataframe after changes.

````python
# code here
total_cranes_visits = birds.loc[birds['birds'] == 'Cranes', 'visits'].sum()
print(f"Total visits of Cranes: {total_cranes_visits}")

# ii. Find the number of each type of birds in dataframe
print("\n--- Count of each bird type ---")
print(birds['birds'].value_counts())

# iii. Print no of duplicate rows
print(f"\nNumber of duplicate rows: {birds.duplicated().sum()}")

# iv. Drop Duplicates rows permanently and show dataframe
birds.drop_duplicates(inplace=True)
print("\n--- DataFrame After Dropping Duplicates ---")
birds
````

###`Q-8:` Team1 vs Team2 Dashbord
Create a function which will take two string(name of two teams) as input. Show win Loss record between them and player getting most player of the match award in matches between these two teams.
```
team1_vs_team2('Kolkata Knight Riders','Chennai Super Kings')
```

````python
def team1_vs_team2(t1, t2):
    # Added outer parentheses around the two main blocks separated by |
    print(ipl['WinningTeam'][((ipl['Team1'] == t1) & (ipl['Team2'] == t2)) | ((ipl['Team2'] == t1) & (ipl['Team1'] == t2))].value_counts())

    print(ipl['Player_of_Match'][((ipl['Team1'] == t1) & (ipl['Team2'] == t2)) | ((ipl['Team2'] == t1) & (ipl['Team1'] == t2))].value_counts().head(1).index[0])

team1_vs_team2('Kolkata Knight Riders', 'Chennai Super Kings')
````
