#### `Q-9:` Write a program implement Boradcasting Rule to check if two array can be added or not.
Given tuples of shapes.
```Python
shape of a- (3,2,2)
shape of b- (2,2)

check_broadcast(a, b) -> return Boolean (True if can broadcasted, False other wise.)
```

````python
def check_broadcast(shape_a, shape_b):
    # Align shapes by padding the shorter one with 1s on the left
    len_a, len_b = len(shape_a), len(shape_b)
    max_len = max(len_a, len_b)

    shape_a = (1,) * (max_len - len_a) + shape_a
    shape_b = (1,) * (max_len - len_b) + shape_b

    # Check compatibility dimension by dimension from right to left
    for dim_a, dim_b in zip(shape_a, shape_b):
        if dim_a != dim_b and dim_a != 1 and dim_b != 1:
            return False
    return True

# Test case
shape_of_a = (3, 2, 2)
shape_of_b = (2, 2)
print(check_broadcast(shape_of_a, shape_of_b))  # Returns True
````
