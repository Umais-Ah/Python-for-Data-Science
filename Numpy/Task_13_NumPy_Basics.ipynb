#### `Q-11:` Softmax function

Create a Python function to calculate the Softmax of the given numpy 1D array. The function only accepts the numpy 1D array, otherwise raise error.

$$\Large \sigma(\vec z)_i = \frac{e^{z_i}}{\sum_{j=i}^{K}{e^{z_j}}}$$

**Example 1:**

Input:
```Python
[86.03331084, 37.7285648, 48.64908087, 87.16563062, 38.40852563, 37.20006318]
```

Output:
```Python
[2.43733249e-01, 2.56112115e-22, 1.41628284e-17, 7.56266751e-01, 5.05514197e-22, 1.50974911e-22]
```

**Example 2:**

Input:
```Python
[33.17344305, 45.61961654, 82.05405781, 80.9647098, 68.82830233, 91.52064278]
```

Output:
```Python
[4.57181035e-26, 1.16249923e-20, 7.73872596e-05, 2.60358426e-05, 1.39571531e-10, 9.99896577e-01]
```

````python
def softmax(array):
  num=np.exp(array)
  den=np.sum(num)
  return num/den


print(softmax(np.array([86.03331084, 37.7285648, 48.64908087, 87.16563062, 38.40852563, 37.20006318])))
````

#### `Q-18` Answer below asked questions on given array:
1. Fetch Every alternate column of the array
2. Normalise the given array

https://en.wikipedia.org/wiki/Normalization_(statistics)

There are different form of normalisation for this question use below formula.

$$\large X_{normalized} = \frac{X - X_{min}}{X_{max} - X_{min}}$$

```python
arr1=np.random.randint(low=1, high=10000, size=40).reshape(8,5)
```

````python
arr1=np.random.randint(low=1, high=10000, size=40).reshape(8,5)
arr2=arr1[:,::2]
def normalization(array):
  return array-np.min(array)/np.max(array)-np.min(array)

print(normalization(arr2))

````

#### `Q-19:` Write a function which will accept 2 arguments.
First: A 1D numpy array arr

Second: An integer n {Please make sure n<=len(arr)}

Output: The output should be the nth largest item out of the array
```Python
Example1 : arr = (12,34,40,7,1,0) and n = 3, the output should be 12
Example2 : arr = (12,34,40,7,1,0) and n = 1, the output should be 40
```

````python
def nth_largest(arr,n):
  sorted=np.sort(arr)[::-1]
  return sorted[n-1]

arr=np.array([12,34,40,7,1,0])
n=1
print(nth_largest(arr,n))
````

#### `Q-20:` Create the following pattern without hardcoding. Use only numpy functions and the below input array a.
```Python
Input: a = np.array([1,2,3])
Output: array([1, 1, 1, 2, 2, 2, 3, 3, 3, 1, 2, 3, 1, 2, 3, 1, 2, 3])
```

````python
a = np.array([1,2,3])
a2=np.hstack((np.repeat(a,3),np.tile(a,3)))
a2
````
